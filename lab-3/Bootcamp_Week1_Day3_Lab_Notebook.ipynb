## Step 4: Data Analysis
- Calculate daily average temperature
- Find max, min, mean for each metric
- Which hour of the day is the most humid on average?

```python
# TODO: Perform analysis
# Use groupby, aggregation, and filtering functions
# Placeholder example:
# df['timestamp'] = pd.to_datetime(df['timestamp'])
# df['hour'] = df['timestamp'].dt.hour
# avg_humidity_by_hour = df.groupby('hour')['humidity_%'].mean()
```

```python
import pandas as pd

df['timestamp'] = pd.to_datetime(df['timestamp'])

# Daily average temperature
print(df.groupby(df['timestamp'].dt.date)['temperature_C'].mean())

# Max, Min, Mean for all numeric metrics
print(df.describe().loc[['min', 'max', 'mean']])

# Most humid hour
df['hour'] = df['timestamp'].dt.hour
hourly_humidity = df.groupby('hour')['humidity_%'].mean()
print(hourly_humidity)
print("Most humid hour:", hourly_humidity.idxmax())
```
