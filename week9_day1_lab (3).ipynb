#Load the Stable Diffusion Model

Stable Diffusion is a pre-trained generative AI model capable of converting text descriptions into images.

In this step, we will load the model and prepare it for image generation.

**Note:** The first execution may take several minutes because the model needs to be downloaded.


Hints:

- Use StableDiffusionPipeline.from_pretrained()
- The model name is:
  runwayml/stable-diffusion-v1-5
- Move the pipeline to the GPU.

```python
#write your code here
import torch
from diffusers import StableDiffusionPipeline

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")
```

# Generate Your First AI Image

Let's generate your first AI image!



Complete the code below to generate an image for the following prompt:

> "A futuristic city at sunset"

Display the generated image.

```python
#write your code here
prompt = "A futuristic city at sunset"

image = pipe(prompt).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image)
plt.axis("off")
plt.show()
```

# Experiment 2: Inference Steps

Stable Diffusion gradually removes noise over multiple iterations.

These iterations are called **Inference Steps**.


Generate the same image using:

- 20 inference steps
- 50 inference steps

```python
#write your code here
prompt = "A futuristic city at sunset"

image20 = pipe(
    prompt,
    num_inference_steps=20
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image20)
plt.axis("off")
plt.show()
```

```python
#write your code here
image50 = pipe(
    prompt,
    num_inference_steps=50
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image50)
plt.axis("off")
plt.show()
```

## Reflection

- Which image looks sharper?

- Was there a noticeable improvement?

- Did image generation take longer?

```python
#answer here
print("""
1. The 50-step image will generally look sharper and more refined.
2. There may be a noticeable improvement, although it depends on the prompt.
3. Yes. 50 inference steps take longer than 20 steps because the model
   performs more denoising iterations.
""")
```

# Experiment 3: Guidance Scale (CFG)

CFG controls how closely Stable Diffusion follows your prompt.


Generate the same prompt using:

- CFG = 5
- CFG = 8
- CFG = 12



```python
#write code here
prompt = "A futuristic city at sunset"

image_cfg5 = pipe(
    prompt,
    guidance_scale=5
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image_cfg5)
plt.axis("off")
plt.show()
```

```python
#write code here
image_cfg8 = pipe(
    prompt,
    guidance_scale=8
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image_cfg8)
plt.axis("off")
plt.show()
```

```python
#write code here
image_cfg12 = pipe(
    prompt,
    guidance_scale=12
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image_cfg12)
plt.axis("off")
plt.show()
```

## Reflection

- Which CFG value produced the best image?

- Which image followed the prompt most accurately?

- What happens when CFG becomes too high?

```python
#answer here
print("""
1. CFG = 8 will generally provide a good balance between image quality
   and prompt adherence.
2. CFG = 8 should follow the prompt accurately while maintaining a
   natural-looking image.
3. When CFG becomes too high, the image can become unnatural, overly
   constrained, distorted, or visually harsh.
""")
```

# Experiment 4: Random Seed

Every image starts from random noise.

Changing the seed changes the starting noise, resulting in different images.



Complete the code below.

Generate images using the following seeds:

- 42
- 100
- 500



```python
#write code here
prompt = "A futuristic city at sunset"

generator = torch.Generator(device="cuda").manual_seed(42)

image42 = pipe(
    prompt,
    generator=generator
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image42)
plt.axis("off")
plt.show()
```

```python
#write code here
generator = torch.Generator(device="cuda").manual_seed(100)

image100 = pipe(
    prompt,
    generator=generator
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image100)
plt.axis("off")
plt.show()
```

```python
#write code here
generator = torch.Generator(device="cuda").manual_seed(500)

image500 = pipe(
    prompt,
    generator=generator
).images[0]

plt.figure(figsize=(8, 8))
plt.imshow(image500)
plt.axis("off")
plt.show()
```

## Reflection

- Which seed generated your favorite image?

- What changed when the seed changed?

- Why are seeds useful when reproducing results?

```python
#answer here
print("""
1. My favorite image was the one generated using the seed that produced
   the most visually appealing result.
2. Changing the seed changed the initial random noise, which resulted
   in different compositions, objects, colors, and details.
3. Seeds are useful because they allow us to reproduce the same image
   when the prompt and generation settings remain the same.
""")
```
