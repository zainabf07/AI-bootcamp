## 2. Quick-start with Pipelines

The `pipeline` API wraps tokenization, model loading, and inference in one object.

### 2.1 Sentiment Analysis

```python
from transformers import pipeline

# 1. Load a sentiment-analysis pipeline (defaults to a small model)
sentiment = pipeline("sentiment-analysis")

# 2. Run inference
examples = [
    "Hugging Face makes NLP super accessible!",
    "I dislike bugs in my code..."
]
results = sentiment(examples)
for text, res in zip(examples, results):
    print(f"{text!r:50} → label={res['label']}, score={res['score']:.3f}")
```

### 2.2 Text Generation

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Once upon a time"
out = generator(prompt, max_length=30, num_return_sequences=1)
print(out[0]["generated_text"])
```

### 2.3 Question Answering

```python
from transformers import pipeline

qa = pipeline("question-answering", model="distilbert-base-cased-distilled-squad")
context = (
    "Hugging Face is an AI company with the mission to democratize good machine learning. "
    "The Transformers library provides thousands of pretrained models in 100+ languages."
)

answer = qa({
    "question": "What is the mission of Hugging Face?",
    "context": context
})
print(answer["answer"])
```

### 2.4 Model Discovery with `huggingface_hub`

```python
from huggingface_hub import HfApi

api = HfApi()
# List top 5 English summarization models
models = api.list_models(task="summarization", limit=5)
for m in models:
    print(m.modelId)
```

# Hugging Face Named Entity Recognition
Objective:

To develop a Named Entity Recognition pipeline using a pre-trained Hugging Face Transformer model and evaluate its performance on different text examples.

Selected Task:

Named Entity Recognition

Selected Model: dslim/bert-base-NER

Library:

 Hugging Face Transformers

Pipeline:

NER / Token Classification

````python
!pip install transformers datasets huggingface_hub --quiet
````

````python
from transformers import pipeline
````

````python
ner = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")
# ner is Named Entity Recognition.
````

````python
text1 = """
Ali is a software engineer working at Microsoft in Lahore.
He specializes in Python and Machine Learning.
"""

results1 = ner(text1)

print("Example 1 Results:")
for entity in results1:
    print(
        f"Entity: {entity['word']:<20} "
        f"Type: {entity['entity_group']:<10} "
        f"Confidence: {entity['score']:.3f}"
    )
````

````python
text2 = """
Dr. Ahmed works for Google in Islamabad.
He is an expert in Artificial Intelligence and Data Science.
"""

results2 = ner(text2)

print("Example 2 Results:")
for entity in results2:
    print(
        f"Entity: {entity['word']:<20} "
        f"Type: {entity['entity_group']:<10} "
        f"Confidence: {entity['score']:.3f}"
    )
````

````python
text3 = """
Muhammad completed his BS Computer Science from FAST University
and joined Systems Limited in Lahore as a software engineer.
"""

results3 = ner(text3)

print("Example 3 Results:")
for entity in results3:
    print(
        f"Entity: {entity['word']:<20} "
        f"Type: {entity['entity_group']:<10} "
        f"Confidence: {entity['score']:.3f}"
    )
````

````python
#pipeline A
ner_simple = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)
````

````python
#pipeline B
ner_first = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="first"
)
````

````python
#comparison
experiment_text = """
Ali works at Microsoft in Lahore as a software engineer.
"""

print("========== SIMPLE ==========")

simple_results = ner_simple(experiment_text)

for entity in simple_results:
    print(entity)


print("\n========== FIRST ==========")

first_results = ner_first(experiment_text)

for entity in first_results:
    print(entity)
````

The pre-trained dslim/bert-base-NER model was successfully implemented using the Hugging Face Transformers pipeline. The model was tested on three professional and technology-related text examples. It identified named entities and assigned entity categories and confidence scores. The experiment also demonstrated how different aggregation strategies can affect the presentation of token-level predictions.
